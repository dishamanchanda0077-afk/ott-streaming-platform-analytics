```python
# ==========================================================
# OTT STREAMING PLATFORM ANALYTICS PROJECT
# DATA CLEANING & PREPARATION SCRIPT
# ==========================================================

import pandas as pd
import numpy as np

print("="*60)
print("LOADING DATASETS")
print("="*60)

users = pd.read_csv("users.csv")
movies = pd.read_csv("movies.csv")
watch_history = pd.read_csv("watch_history.csv")
reviews = pd.read_csv("reviews.csv")
recommendation_logs = pd.read_csv("recommendation_logs.csv")
search_logs = pd.read_csv("search_logs.csv")

datasets = {
    "users": users,
    "movies": movies,
    "watch_history": watch_history,
    "reviews": reviews,
    "recommendation_logs": recommendation_logs,
    "search_logs": search_logs
}

# ==========================================================
# DATASET OVERVIEW
# ==========================================================

print("\nDATASET SHAPES\n")

for name, df in datasets.items():
    print(f"{name}: {df.shape}")

# ==========================================================
# MISSING VALUES
# ==========================================================

print("\n" + "="*60)
print("MISSING VALUE CHECK")
print("="*60)

for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isnull().sum())

# ==========================================================
# DUPLICATES
# ==========================================================

print("\n" + "="*60)
print("DUPLICATE CHECK")
print("="*60)

for name, df in datasets.items():
    print(f"{name}: {df.duplicated().sum()}")

# ==========================================================
# REMOVE DUPLICATES
# ==========================================================

for df in datasets.values():
    df.drop_duplicates(inplace=True)

# ==========================================================
# DATE CONVERSION
# ==========================================================

users['subscription_start_date'] = pd.to_datetime(
    users['subscription_start_date'],
    errors='coerce'
)

users['created_at'] = pd.to_datetime(
    users['created_at'],
    errors='coerce'
)

movies['added_to_platform'] = pd.to_datetime(
    movies['added_to_platform'],
    errors='coerce'
)

watch_history['watch_date'] = pd.to_datetime(
    watch_history['watch_date'],
    errors='coerce'
)

reviews['review_date'] = pd.to_datetime(
    reviews['review_date'],
    errors='coerce'
)

recommendation_logs['recommendation_date'] = pd.to_datetime(
    recommendation_logs['recommendation_date'],
    errors='coerce'
)

search_logs['search_date'] = pd.to_datetime(
    search_logs['search_date'],
    errors='coerce'
)

# ==========================================================
# HANDLE NUMERIC MISSING VALUES
# ==========================================================

movies['imdb_rating'].fillna(
    movies['imdb_rating'].median(),
    inplace=True
)

movies['production_budget'].fillna(
    movies['production_budget'].median(),
    inplace=True
)

movies['box_office_revenue'].fillna(
    movies['box_office_revenue'].median(),
    inplace=True
)

movies['number_of_seasons'].fillna(0, inplace=True)
movies['number_of_episodes'].fillna(0, inplace=True)

watch_history['user_rating'].fillna(
    watch_history['user_rating'].median(),
    inplace=True
)

recommendation_logs['recommendation_score'].fillna(
    recommendation_logs['recommendation_score'].median(),
    inplace=True
)

reviews['helpful_votes'].fillna(0, inplace=True)
reviews['total_votes'].fillna(0, inplace=True)

# ==========================================================
# CREATE ANALYTICAL FEATURES
# ==========================================================

watch_history['completion_flag'] = np.where(
    watch_history['progress_percentage'] >= 90,
    1,
    0
)

recommendation_logs['clicked_flag'] = np.where(
    recommendation_logs['was_clicked'] == True,
    1,
    0
)

search_logs['search_success_flag'] = np.where(
    search_logs['clicked_result_position'].notna(),
    1,
    0
)

reviews['helpfulness_ratio'] = np.where(
    reviews['total_votes'] > 0,
    reviews['helpful_votes'] / reviews['total_votes'],
    0
)

# ==========================================================
# FINAL VALIDATION
# ==========================================================

print("\n" + "="*60)
print("FINAL ROW COUNTS")
print("="*60)

for name, df in datasets.items():
    print(f"{name}: {len(df)}")

# ==========================================================
# EXPORT CLEAN FILES
# ==========================================================

users.to_csv("clean_users.csv", index=False)
movies.to_csv("clean_movies.csv", index=False)
watch_history.to_csv("clean_watch_history.csv", index=False)
reviews.to_csv("clean_reviews.csv", index=False)
recommendation_logs.to_csv("clean_recommendation_logs.csv", index=False)
search_logs.to_csv("clean_search_logs.csv", index=False)

print("\n" + "="*60)
print("EXPORT COMPLETED")
print("="*60)

print("""
Generated Files:

1. clean_users.csv
2. clean_movies.csv
3. clean_watch_history.csv
4. clean_reviews.csv
5. clean_recommendation_logs.csv
6. clean_search_logs.csv
""")
```
